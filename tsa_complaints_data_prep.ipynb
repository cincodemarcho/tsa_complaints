```python
import pandas as pd
import numpy as np

airport = pd.read_csv('complaints-by-airport.csv')
category = pd.read_csv('complaints-by-category.csv')
subcategory = pd.read_csv('complaints-by-subcategory.csv')
iata = pd.read_csv('iata-icao.csv')
```

```python
#filter IATA to US only and rename to match airport column
us_iata = (
    iata[iata["country_code"] == "US"]
    [["iata", "airport", "latitude", "longitude", "region_name"]]
    .rename(columns = {"iata": "airport", "airport": "airport_name", "region_name": "state"}))

#total complaints per airport, then join coordinates
merged = (
    airport
    .groupby("airport", as_index = False)["count"].sum()
    .rename(columns = {"count": "total_complaints"})
    .merge(us_iata, on = "airport", how = "inner"))

#merged.to_csv("merged_choropleth_airports.csv", index = False)
merged.head()
```

```python
iata = (
    iata[iata["country_code"] == "US"]
    [["iata", "airport", "region_name"]]
    .rename(columns = {"iata": "airport", "airport": "airport_name", "region_name": "state"}))

#filter to the two equity categories and pivot to wide format
equity_cats = ["Civil Rights", "Persons w/ Disabilities (PWD)"]

merged2 = (
    category[category["clean_cat"].isin(equity_cats)]
    .groupby(["airport", "clean_cat"], as_index = False)["count"].sum()
    .pivot_table(index = "airport", columns = "clean_cat", values = "count", fill_value = 0)
    .reset_index())

merged2.columns.name = None
merged2 = merged2.rename(columns={"Civil Rights": "civil_rights", "Persons w/ Disabilities (PWD)": "pwd_complaints"})

#add total complaints for bubble sizing
totals = (airport.groupby("airport")["count"].sum().reset_index().rename(columns = {"count": "total_complaints"}))

merged2 = (merged2.merge(totals, on = "airport", how = "left").merge(iata, on = "airport", how = "inner"))

#merged2.to_csv("merged2_equity_airports.csv", index = False)
merged2.head()
```
